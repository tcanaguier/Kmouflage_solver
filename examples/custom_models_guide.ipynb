# Creating custom K(X), coupling A(φ) and potential f(φ) models

This notebook shows how to plug your own kinetic function, conformal
coupling, or scalar potential into the Kinessence solver without touching
`kmouflage/models/`.

`KModel`, `ConformalCoupling` and `Potential` (imported from
`kmouflage.models.*`) are plain dataclasses of callables. The `make_*`
factories already in `kmouflage/models/` (`make_powerlaw_K`,
`make_exponential_coupling`, ...). Any
object with the right fields works identically, wherever it's defined.



```python
import sys
sys.path.insert(0, "..")

import numpy as np
import matplotlib.pyplot as plt

from kmouflage.models.k_functions import KModel
from kmouflage.models.couplings   import ConformalCoupling
from kmouflage.models.potential   import Potential
from kmouflage.solver             import KMouflageBackground, CosmologicalParams
from kmouflage.verify             import verify
```

## 1. Custom K(X)

Kinetic term with an exponential (instead of power-law) correction:

$$K(X) = X + K_0 \left(e^{X/X_\star} - 1\right), \qquad K(0) = 0$$

$$K_X(X) = 1 + \frac{K_0}{X_\star} e^{X/X_\star}, \qquad
  K_{XX}(X) = \frac{K_0}{X_\star^2} e^{X/X_\star}$$

`K`, `Kp`, `Kpp` must work on both a scalar `X` and a numpy array.

The solver asks two things of your K. First K'(X) > 0, so the field has a
healthy kinetic term. Second K'(X) + 2X K''(X) > 0, because it is the
denominator of the Klein-Gordon equation. If one of them reaches zero the run
stops with an error. Keep K(0) = 0, the vacuum energy lives in the potential f(φ).

You don't have to give an initial condition for the field, the solver
computes it from the coupling.

```python
def make_exp_correction_K(K0: float, X_star: float) -> KModel:
    def K(X):
        return X + K0 * (np.exp(np.minimum(X / X_star, 60.0)) - 1.0)

    def Kp(X):
        return 1.0 + (K0 / X_star) * np.exp(np.minimum(X / X_star, 60.0))

    def Kpp(X):
        return (K0 / X_star**2) * np.exp(np.minimum(X / X_star, 60.0))

    return KModel(
        name   = f"exp-correction (K0={K0}, X_star={X_star})",
        K      = K,
        Kp     = Kp,
        Kpp    = Kpp,
        params = {"K0": K0, "X_star": X_star},
    )

custom_K = make_exp_correction_K(K0=1.0, X_star=200.0)
```

## 2. Custom coupling A(φ)

Cubic coupling instead of the built-in linear (exponential) or quadratic
(Gaussian) couplings:

$$A(\phi) = \exp\!\left(\beta\phi + \frac{\kappa}{3}\phi^3\right)$$

$$\alpha(\phi) = \frac{d\ln A}{d\phi} = \beta + \kappa\phi^2, \qquad
  \alpha_{,\phi}(\phi) = \frac{d\alpha}{d\phi} = 2\kappa\phi$$

```python
def make_cubic_coupling(beta: float, kappa: float) -> ConformalCoupling:
    return ConformalCoupling(
        name      = f"cubic (β={beta}, κ={kappa})",
        A         = lambda phi: np.exp(beta * phi + (kappa / 3.0) * phi**3),
        alpha     = lambda phi: beta + kappa * phi**2,
        alpha_phi = lambda phi: 2.0 * kappa * phi,
        params    = {"beta": beta, "kappa": kappa},
    )

custom_coupling = make_cubic_coupling(beta=0.05, kappa=0.01)
```

## 3. Custom potential f(φ) (optional)

`potential=...` is optional. `KMouflageBackground` defaults to `f≡1`
(standard K-mouflage). A custom `Potential` **replaces** that default
entirely rather than adding to it, so if you want a perturbation on top of
the standard model, write it that way explicitly. Here, a Gaussian bump
around the standard f=1:

$$f(\phi) = 1 + f_0\, e^{-(\phi/\phi_\star)^2}, \qquad
  f_{,\phi}(\phi) = -\frac{2\phi}{\phi_\star^2}\,(f(\phi) - 1)$$

```python
def make_gaussian_bump_potential(f0: float, phi_star: float) -> Potential:
    def f(phi):
        return 1.0 + f0 * np.exp(-(phi / phi_star) ** 2)

    def f_phi(phi):
        return -2.0 * phi / phi_star**2 * (f(phi) - 1.0)

    return Potential(
        name   = f"gaussian bump on f=1 (f0={f0}, phi_star={phi_star})",
        f      = f,
        f_phi  = f_phi,
        params = {"f0": f0, "phi_star": phi_star},
    )

custom_potential = make_gaussian_bump_potential(f0=0.1, phi_star=5.0)
```

## 4. Run it

Nothing else changes: `KMouflageBackground` doesn't know or care whether
`model`/`coupling`/`potential` came from `kmouflage.models` or from this
notebook.

```python
bg = KMouflageBackground(
    model     = custom_K,
    coupling  = custom_coupling,
    potential = custom_potential,
    cosmo     = CosmologicalParams(),
)
bg.run()


# verify(bg) à faire plus tard
```
